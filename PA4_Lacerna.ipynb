## Problem C: Category-Average Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt

board2 = pd.read_excel('board2.xlsx')

subjects = ['Math', 'Electronics', 'GEAS', 'Communication']
board2['Average'] = board2[subjects].mean(axis= 1)

# A.)
## doing the summary

### Track

mean_comm = board2.loc[board2['Track'] == 'Communication', 'Average'].mean()
mean_micro = board2.loc[board2['Track'] == 'Microelectronics', 'Average'].mean()
mean_instru = board2.loc[board2['Track'] == 'Instrumentation', 'Average'].mean()

summary_track = pd.DataFrame({'Track' : ['Communication', 'Microelectronics', 'Instrumentation'], 'Mean_average': [mean_comm, mean_micro, mean_instru]})

### Gender

mean_male = board2.loc[board2['Gender'] == 'Male', 'Average'].mean()
mean_female = board2.loc[board2['Gender'] == 'Female', 'Average'].mean()

summary_gender = pd.DataFrame({'Gender' : ['Male', 'Female'], 'Mean_average': [mean_male, mean_female]})

### Hometown

mean_luzon = board2.loc[board2['Hometown'] == 'Luzon', 'Average'].mean()
mean_visayas = board2.loc[board2['Hometown'] == 'Visayas', 'Average'].mean()
mean_mindanao = board2.loc[board2['Hometown'] == 'Mindanao', 'Average'].mean()

summary_hometown = pd.DataFrame({'Hometown' : ['Luzon', 'Visayas', 'Mindanao'], 'Mean_average': [mean_luzon, mean_visayas, mean_mindanao]})

# B.)
## Checking cases
display(summary_track)
print()
display(summary_gender)
print()
display(summary_hometown)
print()

# C.)
## matplotlib in bar graph
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(15, 5))

axes[0].bar(summary_track['Track'], summary_track['Mean_average'])
axes[0].set_title('Mean Average by Track')


axes[1].bar(summary_gender['Gender'], summary_gender['Mean_average'])
axes[1].set_title('Mean Average by Gender')

axes[2].bar(summary_hometown['Hometown'], summary_hometown['Mean_average'])
axes[2].set_title('Mean Average by Hometown')

plt.tight_layout()
plt.show()

# D.)
# identifying the category with the maximum sample mean for each feature
top_track = summary_track.loc[summary_track['Mean_average'].idxmax()]
top_gender = summary_gender.loc[summary_gender['Mean_average'].idxmax()]
top_hometown = summary_hometown.loc[summary_hometown['Mean_average'].idxmax()]

# Checking cases
print(f"1. Track: In the observed dataset, {top_track['Track']} recorded the highest sample mean Average score of {top_track['Mean_average']}.")
print(f"2. Gender: In the observed dataset, {top_gender['Gender']} students achieved the highest sample mean Average score of {top_gender['Mean_average']}.")
print(f"3. Hometown: In the observed dataset, examinees from {top_hometown['Hometown']} obtained the highest sample mean Average score of {top_hometown['Mean_average']}.")


```
